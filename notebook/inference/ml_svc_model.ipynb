```python
import joblib
model_path = "/kaggle/input/models/akashkumbhakar/svc/scikitlearn/default/1/SVC.pkl"
pipeline_path = "/kaggle/input/models/akashkumbhakar/svc/scikitlearn/default/1/StandardScaler.pkl"
model = joblib.load(model_path)
pipeline = joblib.load(pipeline_path)
print("✅ Path to model file:", model_path)
print("✅ Path to pipeline file", pipeline_path)
```
